```python
# Tipos numéricos
edad = 28                 # int
precio = 19.99             # float
numero_complejo = 2 + 3j   # complex

# Texto y valor lógico
nombre = "Lucía"           # str
es_estudiante = True       # bool
sin_dato = None            # NoneType

# Colecciones
colores = ["rojo", "azul"]                    # list
coordenadas = (10, 20)                        # tuple
numeros = range(1, 4)                          # range
persona = {"nombre": nombre, "edad": edad}     # dict
etiquetas = {"python", "datos"}                # set
constantes = frozenset({"pi", "e"})            # frozenset

# Datos binarios
contenido = b"hola"          # bytes
datos_mutables = bytearray(b"hola")  # bytearray

# Mostrar cada variable, su valor y su tipo
variables = [
    edad, precio, numero_complejo, nombre, es_estudiante, sin_dato,
    colores, coordenadas, numeros, persona, etiquetas, constantes,
    contenido, datos_mutables
]

for valor in variables:
    print(f"{valor!r} → {type(valor).__name__}")
```
